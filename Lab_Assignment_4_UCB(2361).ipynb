```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines and plays
M = 3
N_play = 10

# Part B – Initialize N, R, Q and UCB
N = [0, 0, 0]          # Number of times each machine is played
R = [0, 0, 0]          # Total reward received from each machine
Q = [0.0, 0.0, 0.0]    # Estimated Q-values
UCB = [float('inf')] * M

total_reward = 0

print("----- UCB Multi-Armed Bandit -----")

# Part C, D and E – UCB selection, reward collection and Q update
for play in range(1, N_play + 1):

    # Calculate UCB
    for i in range(M):
        if N[i] == 0:
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt((2 * math.log(play)) / N[i])

    # Select machine with highest UCB
    selected = UCB.index(max(UCB))

    print("\nPlay", play)
    print("UCB values:", UCB)
    print("Selected Machine:", machines[selected])

    # Part D – Get reward from user
    reward = float(input("Enter reward: "))

    # Update number of times machine is played
    N[selected] += 1

    # Update total reward
    R[selected] += reward
    total_reward += reward

    # Part E – Incremental sample-average update
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("Updated Q-value:", Q[selected])

# Part F – Display final results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")

# Calculate final UCB values
for i in range(M):
    if N[i] == 0:
        UCB[i] = float('inf')
    else:
        UCB[i] = Q[i] + math.sqrt((2 * math.log(N_play)) / N[i])

    print(f"{machines[i]}\t{N[i]}\t{R[i]:.2f}\t{UCB[i]:.4f}\t\t{Q[i]:.4f}")

print("\nTotal Reward =", total_reward)
print("Total Plays =", N_play)
print("Average Reward =", total_reward / N_play)
```
