```python
import random

# ============================================
# Part A - Machine Initialization
# ============================================

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2


# ============================================
# Part B - Initialize N, R and Q
# ============================================

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

total_reward = 0


# ============================================
# Part C, D and E
# ε-Greedy Decision, Reward and Q Update
# ============================================

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # -----------------------------
    # Exploration
    # -----------------------------
    if r < epsilon:

        selected_index = random.randint(0, M - 1)
        decision = "Exploration"

    # -----------------------------
    # Exploitation
    # -----------------------------
    else:

        max_q = max(Q)

        best_machines = [
            i for i in range(M)
            if Q[i] == max_q
        ]

        selected_index = random.choice(best_machines)
        decision = "Exploitation"

    selected_machine = machines[selected_index]

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Play with machine", selected_machine)

    # ========================================
    # Part D - Reward Collection
    # ========================================

    reward = float(input("Enter reward: "))

    # ========================================
    # Part E - Update N, R and Q
    # ========================================

    N[selected_index] += 1

    R[selected_index] += reward

    Q[selected_index] = (
        Q[selected_index]
        + (1 / N[selected_index])
        * (reward - Q[selected_index])
    )

    total_reward += reward

    print("Updated Q-value:",
          round(Q[selected_index], 4))


# ============================================
# Part F - Display Final Results
# ============================================

print("\n====================================")
print("FINAL RESULTS")
print("====================================")

print("\nMachine\tN\tR\tQ")

for i in range(M):

    print(
        machines[i],
        "\t",
        N[i],
        "\t",
        R[i],
        "\t",
        round(Q[i], 2)
    )

print("\nTotal reward after 10 plays:",
      total_reward)
```
