```python
import gymnasium as gym


def random_agent(env_name, episodes=5):

    # Create environment
    env = gym.make(env_name)

    episode_rewards = []

    print("\n====================================")
    print("Environment:", env_name)
    print("====================================")

    for episode in range(1, episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        done = False

        print("\nEpisode:", episode)
        print("Initial State:", state)

        while not done:

            # Select a random action
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Display information
            print(
                "Action:", action,
                "Reward:", reward,
                "Next State:", next_state
            )

            # Add reward
            total_reward += reward

            # Check whether episode is finished
            done = terminated or truncated

            # Move to next state
            state = next_state

        episode_rewards.append(total_reward)

        print("Total Reward:", total_reward)

    # Calculate average reward
    average_reward = sum(episode_rewards) / len(episode_rewards)

    print("\n------------------------------------")
    print("Results for", env_name)
    print("------------------------------------")
    print("Episode Rewards:", episode_rewards)
    print("Total Reward:", sum(episode_rewards))
    print("Average Reward:", average_reward)

    env.close()


# Run Random Agent on CartPole
random_agent("CartPole-v1", episodes=5)

# Run Random Agent on MountainCar
random_agent("MountainCar-v0", episodes=5)

# Run Random Agent on FrozenLake
random_agent("FrozenLake-v1", episodes=5)
```
