```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# Part B – Initialize knowledge tables
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

total_reward = 0

print("----- Modified Epsilon-Greedy Multi-Armed Bandit -----")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)

# Part C – Modified Epsilon-Greedy Decision
for play in range(1, N_play + 1):

    r = random.random()
    max_Q = max(Q)
    highest_machine = Q.index(max_Q)

    if r < epsilon:
        available = [i for i in range(M) if i != highest_machine]
        selected = random.choice(available)

        decision = "Exploration"

    else:
        selected = highest_machine

        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", machines[selected])

    # Part D – Reward Collection
    reward = float(input("Enter reward: "))

    # Part E – Q-Value Update
    N[selected] += 1
    R[selected] += reward
    Q[selected] = Q[selected] + \
                  (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", [round(x, 2) for x in Q])


# Part F – Display Final Results

print("\n========== FINAL RESULTS ==========")

print("\nParticipation Count (N):")
for i in range(M):
    print("Machine", machines[i], ":", N[i])

print("\nTotal Rewards (R):")
for i in range(M):
    print("Machine", machines[i], ":", R[i])

print("\nEstimated Q-Values:")
for i in range(M):
    print("Machine", machines[i], ":", round(Q[i], 2))

print("\nTotal Reward Received:", total_reward)
best_machine = Q.index(max(Q))

print("\nAnalysis:")
print("Best machine based on estimated Q-value:",
      machines[best_machine])
print("Highest estimated reward:",
      round(Q[best_machine], 2))
```
