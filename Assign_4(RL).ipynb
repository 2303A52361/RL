```python
import math

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

M = 3
N_play = 10

# Part B - Initialize knowledge tables
N = [0, 0, 0]       # Number of times each machine is selected
R = [0, 0, 0]       # Total reward received from each machine
Q = [0.0, 0.0, 0.0] # Estimated reward
UCB = [float('inf'), float('inf'), float('inf')]

total_reward = 0

print("----- UCB Multi-Armed Bandit -----")

for play in range(1, N_play + 1):

    # Part C - Calculate UCB
    for i in range(M):
        if N[i] == 0:
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt((2 * math.log(play)) / N[i])

    # Choose machine with highest UCB
    selected = UCB.index(max(UCB))

    print("\nPlay", play)
    print("UCB Values:")

    for i in range(M):
        if UCB[i] == float('inf'):
            print(machines[i], "= Infinity")
        else:
            print(machines[i], "=", round(UCB[i], 4))

    print("Selected Machine:", machines[selected])

    # Part D - Reward Collection
    reward = float(input("Enter reward: "))

    # Part E - Update N, R and Q
    N[selected] += 1
    R[selected] += reward

    # Incremental sample-average formula
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Updated Q-values:", [round(x, 4) for x in Q])


# Part F - Display Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")

for i in range(M):
    if N[i] == 0:
        final_ucb = float('inf')
    else:
        final_ucb = Q[i] + math.sqrt((2 * math.log(N_play)) / N[i])

    print(
        machines[i],
        "\t", N[i],
        "\t", R[i],
        "\t", "Infinity" if final_ucb == float('inf') else round(final_ucb, 4),
        "\t\t", round(Q[i], 4)
    )

print("\nTotal Reward =", total_reward)
print("Average Reward =", round(total_reward / N_play, 4))
```
